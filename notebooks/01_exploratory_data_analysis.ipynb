### Import libraries and load dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Get the project root (parent directory of notebooks)
PROJECT_ROOT = Path.cwd().parent

# Dataset location
DATA_PATH = PROJECT_ROOT / "data" / "raw" / "online_retail.csv"

# Load dataset
df = pd.read_csv(DATA_PATH)

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", "{:,.2f}".format)

sns.set_theme(style="whitegrid")

print("Dataset loaded successfully!")
print("Project Root:", PROJECT_ROOT)
print("Dataset Path:", DATA_PATH)
print("Dataset Shape:", df.shape)
```

### Dataset overview

```python
print("Dataset Shape:", df.shape)

print("\nFirst 10 Records:")
display(df.head(10))

print("\nLast 5 Records:")
display(df.tail())

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

### Data types and missing values

```python
print("Data Types:")
display(df.dtypes)

missing_values = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df)) * 100
})

missing_values = missing_values.sort_values(
    by="Missing Percentage",
    ascending=False
)

display(missing_values)
```

### Duplicate records

```python
duplicate_count = df.duplicated().sum()

print("Total Duplicate Rows:", duplicate_count)
print("Duplicate Percentage:", round((duplicate_count / len(df)) * 100, 2), "%")
```

### Statistical summary

```python
print("Numerical Column Statistics:")
display(df.describe())

print("\nCategorical Column Statistics:")
display(df.describe(include="object"))
```

### Convert invoice dates

```python
df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"],
    format="%m/%d/%Y %H:%M",
    errors="coerce"
)

print("InvoiceDate converted successfully.")

print("Date Range:")
print("Start:", df["InvoiceDate"].min())
print("End:", df["InvoiceDate"].max())

print("\nInvalid Dates:", df["InvoiceDate"].isna().sum())
```

### Missing values analysis

```python
missing_analysis = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df) * 100).round(2)
})

display(missing_analysis.sort_values(
    by="Missing Count",
    ascending=False
))
```

### Investigate cancelled invoices

```python
df["IsCancelled"] = df["InvoiceNo"].str.startswith("C", na=False)

cancelled_count = df["IsCancelled"].sum()
normal_count = (~df["IsCancelled"]).sum()

print("Cancelled Invoice Records:", cancelled_count)
print("Other Invoice Records:", normal_count)

print("\nCancelled Invoice Examples:")
display(df[df["IsCancelled"]].head(10))
```

### Investigate negative and zero quantities

```python
negative_quantity = (df["Quantity"] < 0).sum()
zero_quantity = (df["Quantity"] == 0).sum()
positive_quantity = (df["Quantity"] > 0).sum()

print("Negative Quantity Records:", negative_quantity)
print("Zero Quantity Records:", zero_quantity)
print("Positive Quantity Records:", positive_quantity)

print("\nNegative Quantity Examples:")
display(df[df["Quantity"] < 0].head(10))
```

### Investigate unit prices

```python
negative_price = (df["UnitPrice"] < 0).sum()
zero_price = (df["UnitPrice"] == 0).sum()
positive_price = (df["UnitPrice"] > 0).sum()

print("Negative UnitPrice Records:", negative_price)
print("Zero UnitPrice Records:", zero_price)
print("Positive UnitPrice Records:", positive_price)

print("\nNegative Price Examples:")
display(df[df["UnitPrice"] < 0].head(10))

print("\nZero Price Examples:")
display(df[df["UnitPrice"] == 0].head(10))
```

### Compare cancellation status with quantity

```python
cancellation_analysis = df.groupby("IsCancelled").agg(
    Total_Records=("InvoiceNo", "count"),
    Negative_Quantity=("Quantity", lambda x: (x < 0).sum()),
    Positive_Quantity=("Quantity", lambda x: (x > 0).sum()),
    Zero_Quantity=("Quantity", lambda x: (x == 0).sum())
)

cancellation_analysis.index = [
    "Normal Invoice" if not x else "Cancelled Invoice"
    for x in cancellation_analysis.index
]

display(cancellation_analysis)
```

### Investigate duplicate records

```python
duplicate_records = df[df.duplicated(keep=False)]

print("Total rows involved in duplicates:", len(duplicate_records))

display(
    duplicate_records
    .sort_values(by=["InvoiceNo", "StockCode"])
    .head(20)
)
```

### Fix the categorical statistics warning

```python
print("Categorical Column Statistics:")

display(df.describe(include=["str"]))
```

### Negative quantities in non-cancelled invoices

```python
non_cancelled_returns = df[
    (~df["IsCancelled"]) & (df["Quantity"] < 0)
]

print("Negative quantity records in non-cancelled invoices:",
      len(non_cancelled_returns))

display(
    non_cancelled_returns[
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "InvoiceDate", "UnitPrice",
         "CustomerID", "Country"]
    ].head(20)
)
```

### Investigate negative unit prices

```python
negative_price_records = df[df["UnitPrice"] < 0]

print("Records with negative prices:", len(negative_price_records))

display(negative_price_records)
```

### Investigate zero unit prices

```python
zero_price_records = df[df["UnitPrice"] == 0]

print("Records with zero prices:", len(zero_price_records))

display(
    zero_price_records[
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "InvoiceDate", "UnitPrice",
         "CustomerID", "Country"]
    ].head(20)
)

print("\nMost frequent zero-price products:")
display(
    zero_price_records["Description"]
    .value_counts()
    .head(15)
)
```

### Examine duplicate records

```python
duplicate_records = df[
    df.duplicated(
        subset=[
            "InvoiceNo", "StockCode", "Description",
            "Quantity", "InvoiceDate", "UnitPrice",
            "CustomerID", "Country"
        ],
        keep=False
    )
]

print("Rows belonging to duplicate groups:", len(duplicate_records))

display(
    duplicate_records
    .sort_values(["InvoiceNo", "StockCode", "InvoiceDate"])
    .head(30)
)
```
